# Фильтр блума

https://www.youtube.com/watch?v=V3pzxngeLqw

Он проверяет входит ли элемент в множество.
При этом может дать точный ответ что элемент не входит.
и не точный ответ о том что входит

false positive - возможен, когда фильтр говорит что элемент есть в множестве но на самом деле элемента нет
false negative - невозможен, когда фильтр говорит что элемента нет а на самом деле он есть.

То есть, если фильтр сказал что в множестве элемента нет - это точный ответ.
Если фильтр сказал что элемент есть - это требует проверки
