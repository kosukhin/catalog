# Shuffle Array

https://leetcode.com/problems/shuffle-the-array/?envType=problem-list-v2&envId=dsa-linear-shoal-array-i

```javascript
(() => {
    var shuffle = function(nums, n) {
        const result = [];
        for(let i = 0; i < n; i++) {
            result.push(nums[i]);
            result.push(nums[i + n])
        }
        return result;
    };

    console.log(shuffle([2,5,1,3,4,7], 3))
})();
```
